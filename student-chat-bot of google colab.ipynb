```python
# AI Chatbot for Student Support Services
# Google Colab Compatible

def chatbot_response(message):
    message = message.lower().strip()

    if any(word in message for word in ["hello", "hi", "hey", "namaste"]):
        return "Hello! 👋 I am your Student Support AI Chatbot. How can I help you?"

    elif "admission" in message:
        return "For admission information, please contact the college admission department."

    elif "fee" in message or "fees" in message:
        return "For fee details, please contact the accounts or fee department."

    elif "exam" in message:
        return "For exam dates and schedules, please contact the examination department."

    elif "attendance" in message:
        return "For attendance details, please contact your respective department."

    elif "library" in message:
        return "The college library provides books, study material and digital resources."

    elif "course" in message:
        return "Please tell me the course name for which you need information."

    elif "time" in message or "timing" in message:
        return "Please check the college timetable or contact your department for class timings."

    elif "result" in message:
        return "For examination results, please check the official college portal."

    elif "thank" in message:
        return "You're welcome! 😊"

    elif "bye" in message:
        return "Goodbye! Have a great day! 👋"

    else:
        return "Sorry, I don't understand your question. Please contact the Student Support Department."


# Chatbot
print("==========================================")
print("   AI CHATBOT FOR STUDENT SUPPORT")
print("==========================================")
print("Type 'bye' to exit the chatbot.\n")

while True:
    user_message = input("You: ")

    response = chatbot_response(user_message)

    print("AI Bot:", response)
    print()

    if user_message.lower().strip() == "bye":
        break
```
